## What is Data?
Data is a collection of facts, observations, measurements, or records that can be analyzed to gain insights.

## Why Data Types Matter
Before analyzing data, Python needs to know what kind of information it is handling.

Examples:
- Age → Integer
- Salary → Float
- Name → String
- Is_Placed → Boolean

These data types determine what operations can be performed.


```python
# Integer (int) - Used for whole numbers.
age = 21
students = 50

print(age)
print(type(age))
```

```python
# Float - Use for decimal values
salary = 55000.75

print(salary)
print(type(salary))
```

```python
# String (str) - stores text
name = "123"

print(name)
print(type(name))
```

```python
# Boolean (bool)- use for true or false
is_placed = True

print(is_placed)
print(type(is_placed))
```

# Python Data Structures

Data analytics rarely works with single values. We often work with collections of data.


## Lists
- Ordered, mutable collections of items.
- Created using square brackets `[]`.
- Allow duplicate elements and mixed data types.


```python
marks = [78, 85, 90, 67, 88]

print(marks)
print(type(marks))
```

```python
sales = [1200, 1500, 1800, 2200]

print(max(sales))
print(min(sales))
print(sum(sales))
```

## Tuples
- Ordered, immutable sequences of items.
- Created using parentheses `()`.
- Cannot be changed after creation.


```python
months = ("Jan", "Feb", "Mar")

print(months)
print(type(months))
```

## Dictionaries (`dict`)
- Mutable key-value collections.
- Created using curly braces `{key: value}`.
- Keys must be unique and hashable.


```python
user = {"name": "Alex", "role": "Developer"}

print(user)
print(type(user))
```

```python
student = {
    "Name": "Aman",
    "Age": 21,
    "CGPA": 8.4
}

print(student)
print(student["CGPA"])
```

## Sets
- Collections of unique elements.
- Created using curly braces `{}` or `set()`.
- Automatically remove duplicates and are useful for membership testing.


```python
unique_ids = {101, 102, 103, 101}
print(unique_ids)
print(type(unique_ids))
```

```python
cities = {"Delhi", "Mumbai", "Delhi", "Pune"}

print(cities)
```

## Practice Question 1
A student is not allowed to sit for an exam if their attendance is less than 75%. Take the number of classes held and attended as input, calculate attendance percentage, and display whether the student is allowed to sit for the exam.


```python
total_classes = int(input('Number of classes held: '))
attended_classes = int(input('Number of classes attended: '))

if total_classes <= 0:
    print('Total classes must be greater than zero.')
elif attended_classes < 0 or attended_classes > total_classes:
    print('Attended classes must be between 0 and total classes held.')
else:
    percentage = (attended_classes / total_classes) * 100
    print(f'Attendance percentage: {percentage:.2f}%')
    if percentage >= 75:
        print('Student is allowed to sit in the exam.')
    else:
        print('Student is not allowed to sit in the exam.')
```

## Practice Question 3
Write a Python function that takes a list and returns a new list containing only unique elements, preserving their original order.


```python
def unique_list(items):
    unique_items = []
    for item in items:
        if item not in unique_items:
            unique_items.append(item)
    return unique_items

values = [1, 2, 3, 3, 3, 3, 4, 5]
print(unique_list(values))
```

## Practice Question 4
Make a two-player Rock-Paper-Scissors game.


```python
choices = {'rock', 'paper', 'scissors'}

while True:
    player1 = input('Enter the first player name: ').strip() or 'Player 1'
    player2 = input('Enter the second player name: ').strip() or 'Player 2'

    choice1 = input(f'{player1}, enter rock/paper/scissors: ').strip().lower()
    choice2 = input(f'{player2}, enter rock/paper/scissors: ').strip().lower()

    if choice1 not in choices or choice2 not in choices:
        print('Invalid input. Please enter rock, paper, or scissors.')
    elif choice1 == choice2:
        print('Match Draw!')
    elif (choice1 == 'rock' and choice2 == 'scissors') or \
         (choice1 == 'paper' and choice2 == 'rock') or \
         (choice1 == 'scissors' and choice2 == 'paper'):
        print(f'{player1} wins!')
    else:
        print(f'{player2} wins!')

    restart = input('Play again? (yes/no): ').strip().lower()
    if restart not in {'yes', 'y'}:
        break
```

## Practice Question 5
Write a Python program to find the items in a list that start with a specific character.


```python
number_of_strings = int(input('How many strings do you want to enter? '))
items = []

for _ in range(number_of_strings):
    items.append(input('Enter a string: '))

letter = input('Enter the starting character to search for: ').strip()
if len(letter) != 1:
    print('Please enter exactly one character.')
else:
    matches = [item for item in items if item.lower().startswith(letter.lower())]
    print('Matching strings:', matches)
    print('Number of matching strings:', len(matches))
```

## Practice Question 10
Write a Python function that takes two lists and returns `True` if they have at least one common member.


```python
list1 = ['red', 'green', 'white', 'black', 'pink', 'yellow']
list2 = ['a', 'b', 'c', 'd', 'red']

def has_common_member(first, second):
    return bool(set(first) & set(second))

print('List 1:', list1)
print('List 2:', list2)
print('Do they have a common member?', has_common_member(list1, list2))
```

## Practice Question 11
Write a program to find the repeated items in a tuple.


```python
from collections import Counter

tuple_items = ('a', 'b', 'c', 'a', 'b', 'd')
counts = Counter(tuple_items)
repeated_items = {item: count for item, count in counts.items() if count > 1}

print('Tuple:', tuple_items)
print('Repeated items:', repeated_items)
```

## Data Pipeline
Raw Data → Import Data → Understand Data → Data Cleaning → Data Transformation → Data Analysis → Visualization → Insights


```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data Cleaning: check missing values
df.isnull().sum()
```

```python
# Fill missing values in Age and Attendance with their respective column means
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mean())
df.head()
```

```python
# Identify duplicate rows
df.duplicated()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
# Standardize capitalization in Gender column
df['Gender'] = df['Gender'].str.title()
```

```python
df.head()
```

# Data Transformation


```python
# Add a new column for performance grade
def grade(cgpa):
    if cgpa >= 9:
        return 'Excellent'
    elif cgpa >= 8:
        return 'Good'
    elif cgpa >= 7:
        return 'Average'
    else:
        return 'Poor'

df['Grade'] = df['CGPA'].apply(grade)
```

```python
df.head()
```

```python
# Create attendance status column
df['Attendance_Status'] = df['Attendance'].apply(
    lambda value: 'Regular' if value >= 75 else 'Irregular'
)
```

```python
df.head()
```

```python
# Basic Data Analysis
print('Mean CGPA:', df['CGPA'].mean())
print('Maximum attendance:', df['Attendance'].max())
print('Placement status counts:')
print(df['Placement_Status'].value_counts())
```

```python
# Attendance distribution
sns.boxplot(y=df['Attendance'])
plt.title('Attendance Boxplot')
plt.show()
```

```python
# Number of students by department
sns.countplot(data=df, x='Department')
plt.title('Students by Department')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Relationship between attendance and CGPA, grouped by placement status
sns.scatterplot(
    data=df,
    x='Attendance',
    y='CGPA',
    hue='Placement_Status'
)
plt.title('Attendance vs CGPA by Placement Status')
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap of numeric columns
plt.figure(figsize=(6, 4))
sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```
